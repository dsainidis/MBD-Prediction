## Creates the **Timeline** for a specific region by using the Grid File and the required dates ##

```python
import pandas as pd
from datetime import date, timedelta

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../../../Modules")

from utils import *

enc = 'utf-8'
dec = 'greek8'
random.seed(0)
```

```python
NUTS0 = 'GR'
NUTS2 = 'Thessaly'
```

```python
nuts2_grid = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Grid_NUTS3.csv', encoding = enc)
nuts2_grid.head()
```

```python
## start and end date is defined from available epidemiological file
```

```python
start = start_dt = date(2024, 7, 2)
end = end_dt = date(2024, 8, 1)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
print(f'Total Rows: {len(dates) * len(nuts2_grid)}')
```

```python
nuts2_timeline = pd.DataFrame()
```

```python
data = []
for index, row in nuts2_grid.iterrows():
    for date in dates:
        data.append({'NUTS2_ID': row['NUTS2_ID'], 
                     'NUTS2_NAME': row['NUTS2_NAME'],
                     'NUTS3_ID': row['NUTS3_ID'], 
                     'NUTS3_NAME': row['NUTS3_NAME'], 
                     'x': row['x'], 
                     'y': row['y'], 
                     'dt_placement': date})

nuts2_timeline = pd.DataFrame(data)
```

```python
nuts2_timeline['dt_placement'] = pd.to_datetime(nuts2_timeline['dt_placement'])
```
