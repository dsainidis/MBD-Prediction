```python
df_loading_gr = df_loading.query("geo.str.match('EL\d{3}')").copy()
```

```python
df_unloading_gr = df_unloading.query("geo.str.match('EL\d{3}')").copy()
```

```python
df_loading_gr = df_loading_gr[df_loading_gr.nst07 == 'TOTAL'].copy()
```

```python
df_unloading_gr = df_unloading_gr[df_unloading_gr.nst07 == 'TOTAL'].copy()
```

```python
df_loading_gr.drop(columns=['DATAFLOW', 'LAST UPDATE', 'freq', 'nst07', 'unit', 'OBS_FLAG'], inplace=True)
```

```python
df_unloading_gr.drop(columns=['DATAFLOW', 'LAST UPDATE', 'freq', 'nst07', 'unit', 'OBS_FLAG'], inplace=True)
```

```python
df_loading_gr.reset_index(drop=True, inplace=True)
```

```python
df_unloading_gr.reset_index(drop=True, inplace=True)
```

```python
df_loading_gr
```

```python
df_unloading_gr
```

```python
df_out_loading = df_loading_gr.rename(columns={"geo": "NUTS3_CODE", "TIME_PERIOD": "YEAR", "OBS_VALUE": "L_TOTAL"}, errors="raise")
```

```python
df_out_unloading = df_unloading_gr.rename(columns={"geo": "NUTS3_CODE", "TIME_PERIOD": "YEAR", "OBS_VALUE": "UNL_TOTAL"}, errors="raise")
```

```python
missingno.matrix(df_out_loading)
```

```python
missingno.matrix(df_out_unloading)
```

```python
df_out = pd.merge(df_out_loading, df_out_unloading,  how='left', on=['NUTS3_CODE', 'YEAR'])
```

```python
missingno.matrix(df_out)
```

```python
nuts3_all = df_out.NUTS3_CODE.unique()
```

```python
df_out.to_csv('../data/GR_EUROSTAT_Trade_Processed.csv', index=False)
```
