```python
import pandas as pd
from datetime import date, timedelta

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../utils")

from plots import *
from string_handling import *

enc = 'utf-8' 

random.seed(0)
np.random.seed(0)

pd.set_option('display.max_columns', None)
```

```python
cm_extnord_grid = pd.read_csv('../../data/ExtNord/CM_ExtNord_Grid.csv', encoding=enc)
cm_extnord_grid.head()
```

```python
start = start_dt = date(2013, 1, 1)
end = end_dt = date(2024, 12, 31)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
print(f'Total Rows: {len(dates) * len(cm_extnord_grid)}')
```

```python
cm_timeline = pd.DataFrame()
```

```python
data = []
for index, row in cm_extnord_grid.iterrows():
    for date in dates:
        data.append({'x': row['x'], 
                     'y': row['y'],
                     'GID0': row['GID0'], 
                     'GID1': row['GID1'],
                     'GID2': row['GID2'], 
                     'GID3': row['GID3'],  
                     'dt_placement': date})
```

```python
cm_timeline = pd.DataFrame(data)
```

```python
cm_timeline
```

```python
cm_timeline['dt_placement'] = pd.to_datetime(cm_timeline['dt_placement'])
cm_timeline['year'] = cm_timeline['dt_placement'].apply(lambda x: x.year)
```

```python
cm_timeline.info()
```

```python
years = cm_timeline.year.unique().tolist()
```

```python
for item in years:
    cm_towrite = cm_timeline[cm_timeline.year == item].drop(columns = ['year'])
    print(f'Year: {item}, Timeline Shape: {cm_towrite.shape}')
    cm_towrite.to_csv(f'../../data/ExtNord/CM_ExtNord_Timeline_{item}.csv', index = False)
```
