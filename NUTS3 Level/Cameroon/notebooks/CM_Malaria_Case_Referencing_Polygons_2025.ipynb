```python
import pandas as pd
import sys

sys.path.insert(0, "../utils")
enc = 'utf-8' 

pd.set_option('display.max_columns', None)
```

```python
cases_df = pd.read_csv('../data/CM_MAL_cases_GID2_2025_fixed.csv', encoding=enc)
polygon_ids = pd.read_csv('../data/CM_WEB_Polygons_processed.csv', encoding=enc)
```

```python
cases_df.drop(columns=['Unnamed: 0'], inplace=True, errors='ignore')
```

```python
cases_df['location'] = list(zip(cases_df['GID0'], cases_df['GID2'], cases_df['GID3']))
```

```python
polygon_ids['location'] = polygon_ids['location'].astype(str)
cases_df['location'] = cases_df['location'].astype(str)
```

```python
referenced_cases = pd.merge(cases_df, polygon_ids, on=['location'], how='left')
```

```python
referenced_cases.drop(columns=['GID0', 'GID2', 'GID3', 'location', 
                               #'Death_UND5', 'Death_OVR5', 'Death_PRGN', 'Death_TOTAL',
                               'Hosp_UND5', 'Hosp_OVR5', 'Hosp_PRGN', 
                               'Sevr_UND5', 'Sevr_OVR5', 'Sevr_PRGN'], inplace = True)
```

```python
referenced_cases = referenced_cases[['id', 'dt_placement', 'Case_UND5', 'Case_OVR5', 'Case_PRGN', 'Case_TOTAL', 'Sevr_TOTAL', 'Hosp_TOTAL']].copy()
```

```python
referenced_cases.rename(columns={'id': 'polygon_id'}, inplace = True)
```

```python
referenced_cases.reset_index(drop=True, inplace=True)
```
