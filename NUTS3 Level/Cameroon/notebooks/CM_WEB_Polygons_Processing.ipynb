```python
import pandas as pd

import random
import numpy as np
import ast

import sys

sys.path.insert(0, "../utils")

from plots import *
from string_handling import *

enc = 'utf-8' 

random.seed(0)
np.random.seed(0)

pd.set_option('display.max_columns', None)
```

```python
file_path = "../data/"
```

```python
cameroon_polygons = pd.read_csv(f"{file_path}CM_WEB_Polygons.csv",  encoding=enc)
```

```python
cameroon_polygons.head()
```

```python
cameroon_polygons['location'] = list(zip(cameroon_polygons['gid0'], cameroon_polygons['gid2'], cameroon_polygons['gid3']))
```

```python
cameroon_polygons.drop(['gid0', 'gid1', 'gid2', 'gid3'], axis=1, inplace=True)
```

```python
cameroon_polygons
```

```python
location_file_path = "locations_from_shapefile.txt"

with open(location_file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

locations = [ast.literal_eval(line.strip().rstrip(',')) for line in lines]

df_locations = pd.DataFrame({'location': locations})
```

```python
df_locations
```

```python
missing_locations = cameroon_polygons[~cameroon_polygons['location'].isin(df_locations['location'])]
missing_locations.location.sort_values().unique().shape
```
