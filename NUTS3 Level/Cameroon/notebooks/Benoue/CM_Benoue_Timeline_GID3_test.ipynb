```python
import pandas as pd
from datetime import date, timedelta

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../../../Modules")

from utils import *

enc = 'utf-8'
random.seed(0)
```

```python
GID0 = 'CM'
GID2_NAME = 'Benoue'
```

```python
gid2_grid = pd.read_csv(f'../../data/{GID2_NAME}/{GID0}_{GID2_NAME}_Grid_GID3.csv', encoding = enc)
gid2_grid.head()
```

```python
start = start_dt = date(2025, 1, 1)
end = end_dt = date(2025, 1, 31)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
print(f'Total Rows: {len(dates) * len(gid2_grid)}')
```

```python
gid2_timeline = pd.DataFrame()
```

```python
data = []
for index, row in gid2_grid.iterrows():
    for date in dates:
        data.append({'GID_2': row['GID_2'], 
                     'GID2_NAME': row['GID2_NAME'],
                     'GID_3': row['GID_3'], 
                     'GID3_NAME': row['GID3_NAME'], 
                     'x': row['x'], 
                     'y': row['y'], 
                     'dt_placement': date})

gid2_timeline = pd.DataFrame(data)
```
