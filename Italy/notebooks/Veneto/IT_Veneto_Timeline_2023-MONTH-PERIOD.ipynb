## Creates the **Timeline** for a specific region by using the Grid File and the required dates ##

```python
import pandas as pd
from datetime import date, timedelta, datetime

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../../Modules")

from utils import *

enc = 'utf-8'

random.seed(0)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
NUTS0 = 'IT'
NUTS2 = 'Veneto'
NUTS2_LOWER = 'veneto'
```

```python
YEAR = 2023
MONTH = 'August'
PERIOD = '1st'
```

```python
nuts2_grid = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Grid.csv', encoding = enc)
nuts2_grid.head()
```

```python
## start and end date is defined from available epidemiological file
```

```python
dt = 1 if PERIOD == '1st' else 16
mn = datetime.strptime(MONTH, '%B').month

start = start_dt = date(YEAR, mn, dt)
end = end_dt = date(YEAR, mn, dt)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
nuts2_timeline = pd.DataFrame()
```

```python
data = []
for index, row in nuts2_grid.iterrows():
    for date in dates:
        data.append({'lau1': row['lau1'], 
                     'region': row['nuts2'], 
                     'x': row['x'], 
                     'y': row['y'], 
                     'dt_placement': date})

nuts2_timeline = pd.DataFrame(data)
```

```python
nuts2_timeline
```

```python
nuts2_timeline['dt_placement'] = pd.to_datetime(nuts2_timeline['dt_placement'])
```

```python
nuts2_timeline.to_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_{YEAR}-{MONTH}-{PERIOD}.csv', index = False)
```
