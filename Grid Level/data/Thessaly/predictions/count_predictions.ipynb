```python
all = len(df)
all
```

```python
count_low = df['risk_score'].between(0, 0.2).sum()
print(count_low)
print(count_low / all *100 )
```

```python
count_low_med = df['risk_score'].between(0.2, 0.4).sum()
print(count_low_med)
print(count_low_med / all *100 )
```

```python
count_med = df['risk_score'].between(0.4, 0.6).sum()
print(count_med)
print(count_med / all *100 )
```

```python
count_high_med = df['risk_score'].between(0.6, 0.8).sum()
print(count_high_med)
print(count_high_med / all *100 )
```

```python
count_high = df['risk_score'].between(0.8, 0.1).sum()
print(count_high)
print(count_high / all *100 )
```

```python
count_low + count_low_med + count_med + count_high_med + count_high == len(df)
```
