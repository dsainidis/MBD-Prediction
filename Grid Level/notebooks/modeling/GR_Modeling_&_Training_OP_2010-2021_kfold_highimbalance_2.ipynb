```python
X = dataset.select_dtypes(exclude=['object']).drop(columns=['cases'])
y = dataset['cases']

train_idx, test_idx, _, _ = train_test_split(X.index, y, test_size = 0.3, random_state = 0, stratify=y)

features_to_remove = ['x', 'y', 'year', 'month']
features_to_remove += ['lc_prop1', 'lc_prop2', 'lc_prop3', 'lc_type2', 'lc_type3', 'lc_type4', 'lc_type5',]
feature_names = X.select_dtypes(exclude=['object']).drop(columns = features_to_remove).columns
print(feature_names.to_list())

print(f"Total Features Used: {len(feature_names)}")
```

```python
data_train = dataset.iloc[train_idx]
data_test = dataset.iloc[test_idx]
```

```python
plot_imbalance(data_train, target_column='cases')
```

```python
plot_imbalance(data_test, target_column='cases')
```

```python
rows_to_remove = data_train.query('cases == 0').sample(n=107000, random_state=0).index

data_train_high = data_train.drop(rows_to_remove)
data_train_high.reset_index(drop=True, inplace=True)

plot_imbalance(data_train_high, target_column='cases')

rows_to_remove = data_train.query('cases == 0').sample(n=203500, random_state=0).index

data_train_low = data_train.drop(rows_to_remove)
data_train_low.reset_index(drop=True, inplace=True)

plot_imbalance(data_train_low, target_column='cases')
```

```python
rows_to_remove = data_test.query('cases == 0').sample(n=87500, random_state=0).index

data_test.drop(rows_to_remove, inplace=True)
data_test.reset_index(drop=True, inplace=True)

plot_imbalance(data_test, target_column='cases')
```

```python
data_train_high = convert_multiple_cases(data_train_high, target_col='cases')
data_train_high.reset_index(inplace = True, drop = True)

sum_negatives = data_train_high['cases'].value_counts().get(0)
sum_positives = data_train_high['cases'].value_counts().get(1)

params = {
    "seed": 0,
    "objective": "binary:logistic",
    "device": "gpu", 
    "eval_metric": "logloss",
    "learning_rate": 0.3,          #default 0.3, range[0,1], large values lead to overfitting
    "min_split_loss": 0.3,          #default 0, range[0,inf], large values lead to underfitting
    "max_depth": 3,                 #default 6, range[0,inf], large values lead to overfitting
    "reg_alpha": 0.5,               #default 0, range[0,inf], large values lead to underfitting
    "reg_lambda": 1.5,              #default 1, range[0,inf], large values lead to underfitting
    "min_child_weight": 1.5,        #default 1, range[0,inf], large values lead to underfitting
    "scale_pos_weight": sum_negatives/sum_positives, 
}

X_train = data_train_high.select_dtypes(exclude=['object']).drop(columns = features_to_remove + ['cases'])
y_train = data_train_high['cases']

scaler_high = MinMaxScaler(feature_range = (-1, 1))
scaler_high.fit(X_train)
X_train_scaled = scaler_high.transform(X_train)
            
imputer_high = KNNImputer()
imputer_high.fit(X_train)
X_train_imputed = imputer_high.transform(X_train_scaled)
    
model = xgb.XGBClassifier(**params)   
trained_model_high = model.fit(X_train_imputed, y_train)
```

```python
data_train_low = convert_multiple_cases(data_train_low, target_col='cases')
data_train_low.reset_index(inplace = True, drop = True)

sum_negatives = data_train_low['cases'].value_counts().get(0)
sum_positives = data_train_low['cases'].value_counts().get(1)

params = {
    "seed": 0,
    "objective": "binary:logistic",
    "device": "gpu", 
    "eval_metric": "logloss",
    "learning_rate": 0.3,          #default 0.3, range[0,1], large values lead to overfitting
    "min_split_loss": 0.3,          #default 0, range[0,inf], large values lead to underfitting
    "max_depth": 3,                 #default 6, range[0,inf], large values lead to overfitting
    "reg_alpha": 0.5,               #default 0, range[0,inf], large values lead to underfitting
    "reg_lambda": 1.5,              #default 1, range[0,inf], large values lead to underfitting
    "min_child_weight": 1.5,        #default 1, range[0,inf], large values lead to underfitting
    "scale_pos_weight": sum_negatives/sum_positives, 
}

X_train = data_train_low.select_dtypes(exclude=['object']).drop(columns = features_to_remove + ['cases'])
y_train = data_train_low['cases']

scaler_low = MinMaxScaler(feature_range = (-1, 1))
scaler_low.fit(X_train)
X_train_scaled = scaler_low.transform(X_train)
            
imputer_low = KNNImputer()
imputer_low.fit(X_train)
X_train_imputed = imputer_low.transform(X_train_scaled)
    
model = xgb.XGBClassifier(**params)   
trained_model_low = model.fit(X_train_imputed, y_train)
```

```python
data_test['cases'] = data_test['cases'].apply(lambda x: 1 if x > 0 else 0)
data_test.reset_index(inplace = True, drop = True)

X_test = data_test.select_dtypes(exclude=['object']).drop(columns = features_to_remove + ['cases'])
y_test = data_test['cases']
```

```python
X_test_scaled_high = scaler_high.transform(X_test)            
X_test_imputed_high = imputer_high.transform(X_test_scaled_high)

X_test_scaled_low = scaler_low.transform(X_test)            
X_test_imputed_low = imputer_low.transform(X_test_scaled_low)
    
y_test_prob_high = trained_model_high.predict_proba(X_test_imputed_high)
y_test_prob_low = trained_model_low.predict_proba(X_test_imputed_low)
```

```python
df_test_high = pd.DataFrame({'score': y_test_prob_high[:, 1], 'target': y_test})
df_test_low = pd.DataFrame({'score': y_test_prob_low[:, 1], 'target': y_test})

test_loss_high = log_loss(y_test, y_test_prob_high[:, 1])
print(f"Test Loss High: {test_loss_high:.4f}")

test_loss_low = log_loss(y_test, y_test_prob_low[:, 1])
print(f"Test Loss Low: {test_loss_low:.4f}")
```
