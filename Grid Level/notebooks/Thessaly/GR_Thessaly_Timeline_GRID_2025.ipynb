## Creates the **Timeline** for a specific region by using the Grid File and the required dates ##

```python
import pandas as pd
from datetime import date, timedelta

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../../Modules")

from utils import *

pd.options.display.max_columns = None

enc = 'utf-8'
dec = 'greek8'
random.seed(0)
```

```python
NUTS0 = 'GR'
NUTS2 = 'Thessaly'
```

```python
gr_grid = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_GRID.csv', encoding = enc)
gr_grid
```

```python
## start and end date is defined from available epidemiological file
```

```python
start = start_dt = date(2025, 5, 2)
end = end_dt = date(2025, 6, 1)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
gr_timeline = pd.DataFrame()
```

```python
data = []
for index, row in gr_grid.iterrows():
    for date in dates:
        data.append({'x': row['x'], 
                     'y': row['y'], 
                     'cell_number': row['cell_number'],  
                     'dt_placement': date,
                     'NUTS2_ID': row['NUTS2_ID'],
                     })

gr_timeline = pd.DataFrame(data)
```

```python
gr_timeline
```

```python
gr_timeline['dt_placement'] = pd.to_datetime(gr_timeline['dt_placement'])
```

```python
gr_timeline['year'] = gr_timeline['dt_placement'].apply(lambda x: x.year)
```

```python
gr_timeline.info()
```

```python
years = gr_timeline.year.unique().tolist()
```

```python
for item in years:
    gr_timeline[gr_timeline.year == item].drop(columns = ['year']).to_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_GRID_{item}.csv', index = False)
```
