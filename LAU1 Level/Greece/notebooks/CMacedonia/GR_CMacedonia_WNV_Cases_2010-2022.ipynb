## Creates the **Epidemiological Timeline** for a specific region by using the available data on WNV Cases ##

```python
import pandas as pd

enc = 'utf-8'
dec = 'greek8'

pd.options.display.max_columns = None
pd.options.display.max_rows = 10
```

```python
NUTS0 = 'GR'
NUTS2 = 'CMacedonia'
NUTS2_EL = 'κεντρικης μακεδονιας'
```

## 2010 - 2021 ##

```python
data_2021 = pd.read_csv(f'../../data/{NUTS0}_WNV_cases_2010-2021_processed.csv', encoding = enc)
data_2021.head(5)
```

```python
df_21 = data_2021[data_2021['nuts2'].isin([NUTS2_EL])].copy()
```

```python
df_21.reset_index(drop = True, inplace = True)
```

```python
df_21
```

```python
wnv_lau1_list_21 = df_21['lau1'].drop_duplicates().sort_values().tolist()

with open(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_WNV_2010-2021_LAU1_Units_Processed.txt', 'w', encoding=enc) as f:
    for item in wnv_lau1_list_21:
        f.write("%s\n" % item)
  
print(f"Number of LAU1 Units in {NUTS2} (from cases): {len(wnv_lau1_list_21)}")
```

```python
df_21['period'] = df_21.apply(lambda row: 1 if (row['day']>=1 and row['day']<=15) else 2, axis = 1)
df_21['dt_placement'] = df_21.apply(lambda row: f"{row['year']:04}-{row['month']:02}-{1 if row['period'] == 1 else 16:02}", axis = 1)

df_21['dt_placement']= pd.to_datetime(df_21['dt_placement'])
df_21['case'] = 1
```

```python
df_21
```

```python
df_21.drop(columns = ['onset of symptoms', 'day', 'nuts3', 'lau2', 'age', 'sex', 'period'], inplace = True)
```

```python
df_21 = df_21[['month', 'year', 'lau1', 'nuts2', 'dt_placement', 'case']]
```

```python
df_21
```

```python
case_months = df_21['month'].drop_duplicates().sort_values().tolist()
case_months
```

```python
case_years = df_21['year'].drop_duplicates().sort_values().tolist()
case_years
```

```python
with open(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_WNV_2010-2021_Cases_Dates.txt', 'w', encoding=enc) as f:

    f.write("%s: " % 'months')
    for item in case_months:
        f.write("%s, " % item)

    f.write("\n")

    f.write("%s: " % 'years')
    for item in case_years:
        f.write("%s, " % item)
```

```python
df_21.reset_index(inplace = True, drop=True)
```

## 2022 ##

```python
data_2022 = pd.read_csv(f'../../data/{NUTS0}_WNV_cases_2022_processed.csv', encoding = enc)
data_2022.head(5)
```

```python
data_2022.shape
```

```python
df_22 = data_2022[data_2022['nuts2'].isin([NUTS2_EL])].copy()
df_22.reset_index(drop = True, inplace = True)
df_22
```

```python
wnv_lau1_list_22 = df_22['lau1'].drop_duplicates().sort_values().tolist()

with open(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_WNV_2022_LAU1_Units_Processed.txt', 'w', encoding=enc) as f:
    for item in wnv_lau1_list_22:
        f.write("%s\n" % item)
  
print(f"Number of LAU1 Units in {NUTS2} (from cases): {len(wnv_lau1_list_22)}")
```

```python
df_22['period'] = df_22.apply(lambda row: 1 if (row['day']>=1 and row['day']<=15) else 2, axis = 1)
df_22['dt_placement'] = df_22.apply(lambda row: f"{row['year']:04}-{row['month']:02}-{1 if row['period'] == 1 else 16:02}", axis = 1)

df_22['dt_placement']= pd.to_datetime(df_22['dt_placement'])
df_22
```

```python
df_22.drop(columns = ['onset of symptoms', 'day', 'nuts3', 'period'], inplace = True)
df_22 = df_22[['month', 'year', 'lau1', 'nuts2', 'dt_placement', 'cases']]
df_22
```

```python
df_22.cases.sum()
```

```python
case_months = df_22['month'].drop_duplicates().sort_values().tolist()
case_months
```

```python
case_years = df_22['year'].drop_duplicates().sort_values().tolist()
case_years
```

```python
with open(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_WNV_2022_Cases_Dates.txt', 'w', encoding=enc) as f:

    f.write("%s: " % 'months')
    for item in case_months:
        f.write("%s, " % item)

    f.write("\n")

    f.write("%s: " % 'years')
    for item in case_years:
        f.write("%s, " % item)
```

```python
df_22.reset_index(inplace = True, drop=True)
```

```python
## Creating DataFrame to hold negative examples
municipalities = wnv_lau1_list_21 + wnv_lau1_list_22
municipalities_set = set(municipalities)
municipalities_sorted = sorted(municipalities_set)


df_timeline = pd.DataFrame(columns = ['lau1', 'nuts2', 'dt_placement', 'case'])

for municipality in municipalities_sorted:
    for year in range(2010, 2023):
        for month in range (6,12):
            df_timeline = df_timeline.append({'lau1' : municipality, 'nuts2' : NUTS2_EL, 'dt_placement' : f"{year:04}-{month:02}-{1:02}", 'case' : 0}, ignore_index = True)
            df_timeline = df_timeline.append({'lau1' : municipality, 'nuts2' : NUTS2_EL, 'dt_placement' : f"{year:04}-{month:02}-{16:02}", 'case' : 0}, ignore_index = True)
            
df_timeline 
```

```python
df_timeline['dt_placement']= pd.to_datetime(df_timeline['dt_placement'])
df_timeline.reset_index(inplace = True, drop=True)
```

```python
## Add cases to negative examples DataFrame

#iterrate DataFrame with cases by row
for index, row in df_21.iterrows():
    #find index at DataFrame with negative examples where location and date matches the DataFrame with cases
    idx = df_timeline.index[(df_timeline['lau1']==row['lau1']) & (df_timeline['dt_placement']==row['dt_placement'])]
    #get the previous case number using index
    previous_case = df_timeline._get_value(idx.values[0], 'case')
    #add 1 (one) to column case
    df_timeline.at[idx.values[0],'case'] = previous_case+1
```

```python
df_timeline.case.sum()
```

```python
## Add cases to negative examples DataFrame

#iterrate DataFrame with cases by row
for index, row in df_22.iterrows():
    #find index at DataFrame with negative examples where location and date matches the DataFrame with cases
    idx = df_timeline.index[(df_timeline['lau1']==row['lau1']) & (df_timeline['dt_placement']==row['dt_placement'])]
    #get the previous case number using index
    try:
        previous_case = df_timeline._get_value(idx.values[0], 'case')
    except IndexError:
        print(index)
        print(idx)
    current_case = row['cases']
    #add 1 (one) to column case
    df_timeline.at[idx.values[0],'case'] = previous_case+current_case
```

```python
df_timeline.case.sum()
```

```python
df_timeline.to_csv(f"../../data/{NUTS2}/{NUTS0}_{NUTS2}_WNV_Cases_2010-2022.csv", encoding = enc, index = False)
```
