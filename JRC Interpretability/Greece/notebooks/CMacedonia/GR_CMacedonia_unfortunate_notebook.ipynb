```python
import pandas as pd
import missingno

enc = 'utf-8'
```

```python
NUTS0 = 'GR'
NUTS2 = 'CMacedonia'
```

```python
timeline = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_NUTS3_2010-2022.csv', encoding=enc)
timeline_filled_mean = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_NUTS3_2010-2022_filled_meantemps.csv', encoding=enc)
```

```python
timeline
```

```python
timeline_filled_mean
```

```python
timeline_filled_mean.drop(columns=['nuts2'], inplace = True)
timeline_filled_mean.rename(columns={"nuts3": "NUTS3_NAME"}, inplace=True)
```

```python
timeline.NUTS2_NAME.unique()
```

```python
timeline_filled_mean.NUTS3_NAME.unique()
```

```python
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('ημαθιας','ημαθια'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('θεσσαλονικης','θεσσαλονικη'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('πελλας','πελλα'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('πιεριας','πιερια'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('σερρων','σερρες'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('χαλκιδικης','χαλκιδικη'))
```

```python
timeline_filled_mean
```

```python
nuts3_id = {'ημαθια': 'EL521', 
            'θεσσαλονικη': 'EL522', 
            'κιλκις': 'EL523', 
            'πελλα': 'EL524', 
            'πιερια': 'EL525',
            'σερρες': 'EL526',
            'χαλκιδικη': 'EL527'}

```

```python
timeline_filled_mean['NUTS3_ID'] = timeline_filled_mean['NUTS3_NAME'].map(nuts3_id)
timeline_filled_mean['NUTS2_ID'] = 'EL52'
timeline_filled_mean['NUTS2_NAME'] = 'κεντρικη μακεδονια'
```
