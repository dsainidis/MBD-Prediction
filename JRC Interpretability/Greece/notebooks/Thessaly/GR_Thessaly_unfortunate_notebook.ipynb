```python
import pandas as pd
import missingno

enc = 'utf-8'
```

```python
NUTS0 = 'GR'
NUTS2 = 'Thessaly'
```

```python
timeline = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_NUTS3_2010-2022.csv', encoding=enc)
timeline_filled_mean = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_NUTS3_2010-2022_filled_meantemps.csv', encoding=enc)
```

```python
timeline_filled_mean.drop(columns=['nuts2'], inplace = True)
timeline_filled_mean.rename(columns={"nuts3": "NUTS3_NAME"}, inplace=True)
```

```python
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('καρδιτσας','καρδιτσα, τρικαλα'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('λαρισας','λαρισα'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('μαγνησιας','μαγνησια, σποραδες'))
timeline_filled_mean['NUTS3_NAME'] = timeline_filled_mean['NUTS3_NAME'].apply(lambda x : x.replace('τρικαλων','καρδιτσα, τρικαλα'))
```

```python
timeline_filled_mean
```

```python
nuts3_id = {'καρδιτσα, τρικαλα': 'EL611', 
            'λαρισα': 'EL612', 
            'μαγνησια, σποραδες': 'EL613' 
        }
```

```python
timeline_filled_mean['NUTS3_ID'] = timeline_filled_mean['NUTS3_NAME'].map(nuts3_id)
timeline_filled_mean['NUTS2_ID'] = 'EL61'
timeline_filled_mean['NUTS2_NAME'] = 'θεσσαλια'
```
