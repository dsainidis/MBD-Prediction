```python
import numpy as np
import matplotlib.pyplot as plt

def create_horizontal_stacked_barchart(data_list, feature_names, label_list):
    """
    Create a horizontal stacked bar chart.

    Parameters:
    - data_list: List of numpy 1d arrays representing bars.
    - feature_names: List of feature names for each numpy 1d array.
    - label_list: List of labels for the bars.

    Returns:
    - None (displays the plot).
    """
    num_features = len(data_list)
    num_bars = len(data_list[0])

    # Check if the lengths of all arrays are the same
    if any(len(arr) != num_bars for arr in data_list):
        raise ValueError("All numpy arrays must have the same length.")

    # Sort the bars by length (sum of values in each array)
    sorted_indices = np.argsort(np.sum(data_list, axis=1))

    # Create a horizontal stacked bar chart
    fig, ax = plt.subplots(figsize=(10, 6))

    for i in reversed(sorted_indices):
        feature_data = data_list[i]
        bottom_values = np.sum(data_list[:i], axis=0) if i > 0 else 0
        ax.barh(label_list, feature_data, left=bottom_values, label=feature_names[i])

    ax.set_xlabel('Values')
    ax.set_ylabel('Labels')
    ax.set_title('Horizontal Stacked Bar Chart')
    ax.legend()

    plt.show()
```

```python
# Example usage:
data1 = np.array([3, 5, 2, 8])
data2 = np.array([2, 6, 3, 5])
data3 = np.array([1, 4, 2, 7])


b = ['ft1', 'ft2', 'ft3', 'ft4']


a = ['cls1', 'cls2', 'cls3']



create_horizontal_stacked_barchart([data1, data2, data3], a, b)
```
