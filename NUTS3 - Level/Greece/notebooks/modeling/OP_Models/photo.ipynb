```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df_1 = pd.read_table(r'C:\Users\dimit\Downloads\uv_flash_light.txt', header=None, names= ['wavelength', 'photoluminescence'])
df_2 = pd.read_table(r'C:\Users\dimit\Downloads\top_com380nm_flash_light.txt', header=None, names= ['wavelength', 'photoluminescence'])
df_3 = pd.read_table(r'C:\Users\dimit\Downloads\top_com365nm_flash_light.txt', header=None, names= ['wavelength', 'photoluminescence'])
df_4 = pd.read_table(r'C:\Users\dimit\Downloads\boruit_flash_light.txt', header=None, names= ['wavelength', 'photoluminescence'])
```

```python
plt.figure(figsize=(20, 12))
plt.scatter(df_1['wavelength'], df_1['photoluminescence'], c=df_1['wavelength'], cmap='viridis')

# Adding labels and title
plt.xticks(np.arange(200,570,10), rotation=45)
plt.xlabel('Wavelength', size = 20)
plt.ylabel('Photoluminescence', size = 20)
plt.yscale('log')

# Adding color bar
plt.colorbar(label='Photoluminescence')
plt.grid(True)

# Display the plot
plt.show()
```
