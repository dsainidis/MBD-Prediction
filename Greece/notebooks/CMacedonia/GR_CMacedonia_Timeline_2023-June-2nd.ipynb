## Creates the **Timeline** for a specific region by using the Grid File and the required dates ##

```python
import pandas as pd
from datetime import date, timedelta

import random
import geopandas as gpd
import numpy as np

import sys

sys.path.insert(0, "../../../Modules")

from preprocessing_utils import *

enc = 'utf-8'
dec = 'greek8'
random.seed(0)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
NUTS0 = 'GR'
NUTS2 = 'CMacedonia'
NUTS2_EL = 'κεντρικης μακεδονιας'
```

```python
MONTH = 'June'
PERIOD = '2nd'
```

```python
nuts2_grid = pd.read_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Grid.csv', encoding = enc)
nuts2_grid.head()
```

```python
## start and end date is defined from available epidemiological file
```

```python
start = start_dt = date(2023, 6, 16)
end = end_dt = date(2023, 6, 16)
delta = timedelta(days=1)
```

```python
dates = []

while start_dt <= end_dt:
    dates.append(start_dt.isoformat())
    start_dt += delta
```

```python
print('Dates between', start, 'and', end)
print(dates)
```

```python
len(dates)
```

```python
nuts2_timeline = pd.DataFrame()
```

```python
data = []
for index, row in nuts2_grid.iterrows():
    for date in dates:
        data.append({'lau1': row['lau1'], 
                     'region': row['nuts2'], 
                     'x': row['x'], 
                     'y': row['y'], 
                     'dt_placement': date})

nuts2_timeline = pd.DataFrame(data)
```

```python
nuts2_timeline
```

```python
nuts2_timeline['dt_placement'] = pd.to_datetime(nuts2_timeline['dt_placement'])
```

```python
nuts2_timeline.info()
```

```python
nuts2_timeline.to_csv(f'../../data/{NUTS2}/{NUTS0}_{NUTS2}_Timeline_2023-{MONTH}-{PERIOD}.csv', index = False)
```
