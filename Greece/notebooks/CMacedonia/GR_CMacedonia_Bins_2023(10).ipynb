```python
a_list = [124.3636364, 111.9272727, 99.49090909, 87.05454545, 74.61818182, 62.18181818, 49.74545455, 37.30909091, 24.87272727, 12.43636364]
```

```python
import math
indices = []

# Loop through the elements in the float_list and get the floor of each element
for num in a_list:
    floor_value = math.floor(num)
    indices.append(floor_value)
```

```python
indices
```

```python
bins = []
sum = 0
for index in indices:
    sum = sum + index
    bin_lim = concated_sorted.loc[sum, 'score']
    bins.append(bin_lim)

bins.append(1)
```

```python
bins[0] = 0
```

```python
bins
```

```python
concated_og['risk_class'] = pd.cut(concated_og['score'], bins=bins, labels=False, include_lowest=True)
```

```python
concated_og.risk_class.value_counts().sort_index()
```

```python
bins_formatted = [ '%.4f' % elem for elem in bins]
print(bins_formatted)
```
