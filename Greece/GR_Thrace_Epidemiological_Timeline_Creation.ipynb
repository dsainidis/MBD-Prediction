```python
import pandas as pd

enc = 'utf-8'
dec = 'greek8'

pd.options.display.max_columns = None
pd.options.display.max_rows = 10
```

```python
data = pd.read_csv('data/GR_WNV_cases_2010-2021_processed.csv', encoding = enc)
data.head(5)
```

```python
df_thrace = data[data['nuts2'].isin(['ανατολικης μακεδονιας & θρακης'])].copy()
```

```python
df_thrace.reset_index(drop = True, inplace = True)
```

```python
df_thrace
```

```python
thrace_lau1 = df_thrace['lau1'].drop_duplicates().sort_values().tolist()

with open('data/GR_WNV_Thrace_lau1_processed.txt', 'w', encoding=enc) as f:
    for item in thrace_lau1:
        f.write("%s\n" % item)
  
print(f"Number of LAU1 Units in Thrace (from cases): {len(thrace_lau1)}")
```

```python
df_thrace['period'] = df_thrace.apply(lambda row: 1 if (row['day']>=1 and row['day']<=15) else 2, axis = 1)
df_thrace['dt_placement'] = df_thrace.apply(lambda row: f"{row['year']:04}-{row['month']:02}-{1 if row['period'] == 1 else 16:02}", axis = 1)

df_thrace['dt_placement']= pd.to_datetime(df_thrace['dt_placement'])
df_thrace['case'] = 1
```

```python
df_thrace
```

```python
df_thrace.drop(columns = ['onset of symptoms', 'day', 'nuts3', 'lau2', 'age', 'sex', 'period'], inplace = True)
```

```python
df_thrace = df_thrace[['month', 'year', 'lau1', 'nuts2', 'dt_placement', 'case']]
```

```python
df_thrace
```

```python
df_thrace['month'].drop_duplicates().sort_values().tolist()
```

```python
df_thrace['year'].drop_duplicates().sort_values().tolist()
```

```python
## Creating DataFrame to hold negative examples

df_timeline = pd.DataFrame(columns = ['lau1', 'nuts2', 'dt_placement', 'case'])

for municipality in df_thrace['lau1'].drop_duplicates().sort_values():
    for year in range(2010, 2022):
        for month in range (6,11):
            df_timeline = df_timeline.append({'lau1' : municipality, 'nuts2' : 'ανατολικης μακεδονιας και θρακης', 'dt_placement' : f"{year:04}-{month:02}-{1:02}", 'case' : 0}, ignore_index = True)
            df_timeline = df_timeline.append({'lau1' : municipality, 'nuts2' : 'ανατολικης μακεδονιας και θρακης', 'dt_placement' : f"{year:04}-{month:02}-{16:02}", 'case' : 0}, ignore_index = True)
            
df_timeline 
```

```python
df_timeline['dt_placement']= pd.to_datetime(df_timeline['dt_placement'])
```

```python
df_thrace.reset_index(inplace = True, drop=True)
df_timeline.reset_index(inplace = True, drop=True)
```

```python
## Add cases to negative examples DataFrame

#iterrate DataFrame with cases by row
for index, row in df_thrace.iterrows():
    #find index at DataFrame with negative examples where location and date matches the DataFrame with cases
    idx = df_timeline.index[(df_timeline['lau1']==row['lau1']) & (df_timeline['dt_placement']==row['dt_placement'])]
    #get the previous case number using index
    previous_case = df_timeline._get_value(idx.values[0], 'case')
    #add 1 (one) to column case
    df_timeline.at[idx.values[0],'case'] = previous_case+1
```
