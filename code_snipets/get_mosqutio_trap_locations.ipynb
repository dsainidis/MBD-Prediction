```python
import pandas as pd
import json

import random
import geopandas as gpd
import numpy as np
import os
from shapely.geometry import Point

enc = 'utf-8'
pd.set_option('display.max_columns', None)
random.seed(0)
np.random.seed(0)
```

```python
df = pd.read_csv('../data_snipets/GRE_culex_2011_2023.csv', encoding=enc)
```

```python
df.head()
```

```python
df_sub = df.loc[df['nuts2_code'] == 'EL52'].copy()
df_sub.reset_index(drop=True, inplace=True)
```

```python
df_sub.head()
```

```python
df_locations = df_sub[['x', 'y']].drop_duplicates()
df_locations.reset_index(drop=True, inplace=True)
```

```python
df_locations
```

```python
geometry = [Point(xy) for xy in zip(df_locations['x'], df_locations['y'])]
gdf = gpd.GeoDataFrame(df_locations, geometry=geometry)
gdf.set_crs(epsg=4326, inplace=True)  # Assuming WGS84 (lat/lon)
```

```python
gdf.to_file('../data_snipets/EL52_mosquito_traps.shp', encoding = enc)
```
